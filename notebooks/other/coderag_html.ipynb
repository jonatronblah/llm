```python
tokenizer = AutoTokenizer.from_pretrained("meta-llama/Meta-Llama-3.1-8B")
Settings.embed_model = HuggingFaceEmbedding(
    model_name="BAAI/bge-small-en-v1.5"
)

set_global_tokenizer(
    tokenizer.encode
)
```

```python
r = httpx.get("https://memory-alpha.fandom.com/wiki/Jean-Luc_Picard/")
html_text = r.text
soup = BeautifulSoup(html_text)
tags = [tag.name for tag in soup.find_all()]
html_doc = Document(text=html_text)
```

```python
tags = ["p", "h1", "h2", "h3", "h4", "h5", "h6", "li", "b", "i", "u", "section"]
parser = HTMLNodeParser(tags=tags)
nodes = parser.get_nodes_from_documents([html_doc])
print(len(nodes))
```

```python
vector_store = PGVectorStore.from_params(
    database='grover',
    host='postgres',
    password='grover',
    port=5432,
    user='grover',
    table_name="jeanluc",
    embed_dim=384,  
    hnsw_kwargs={
        "hnsw_m": 16,
        "hnsw_ef_construction": 64,
        "hnsw_ef_search": 40,
        "hnsw_dist_method": "vector_cosine_ops",
    },
)
storage_context = StorageContext.from_defaults(vector_store=vector_store)
```

```python
index = VectorStoreIndex(nodes, storage_context=storage_context, show_progress=True)
```
