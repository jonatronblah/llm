```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

```

```python
tokenizer = AutoTokenizer.from_pretrained("Zhengyi/LLaMA-Mesh", device_map = 'cuda')
model = AutoModelForCausalLM.from_pretrained("Zhengyi/LLaMA-Mesh", device_map = 'cuda')
```

```python
prompt = "reate a 3D obj file that includes both vertices and faces using the following description: a truck."
inputs = tokenizer(prompt, return_tensors="pt").to('cuda')
```

```python
outputs = model.generate(**inputs, max_length=3000)
mesh_data = tokenizer.decode(outputs[0], skip_special_tokens=True)
```
